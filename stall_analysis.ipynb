## Introduction

This project analyses 14 days of daily sales and stock records for Wanjiku, a fresh-produce stall owner at Gikomba Market. The objective is to evaluate business performance, identify sales trends, compare weekday and weekend performance, identify products requiring restocking, and provide practical business recommendations using core Python.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall Sales Performance

In this task, I calculate the total, average, highest and lowest sales over the 14-day period. I also identify the corresponding day names and weeks to understand the stall's overall performance.

```python
# Calculate overall sales statistics

total_sales = sum(daily_sales)

average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)

lowest_sales = min(daily_sales)


# Identify the positions of the highest and lowest sales

best_index = daily_sales.index(highest_sales)

worst_index = daily_sales.index(lowest_sales)


# Identify the day names

best_day = days[best_index]

worst_day = days[worst_index]


# Identify the week numbers

best_week = (best_index // 7) + 1

worst_week = (worst_index // 7) + 1


# Display the results

print("OVERALL SALES PERFORMANCE")
print("-" * 35)

print(f"Total Sales: KES {total_sales:,.2f}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,.2f}")
print(f"Lowest Sales: KES {lowest_sales:,.2f}")

print(f"Best Day: {best_day}, Week {best_week}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales and Growth

In this task, I create a reusable function called week_totals() that calculates the sales totals for the first and second weeks using list slicing. I then calculate the percentage growth between the two weeks.

```python
# Function to calculate weekly sales totals

def week_totals(sales):
    """
    Calculate and return the sales totals for week 1 and week 2.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2


# Call the function and unpack the results

week1_total, week2_total = week_totals(daily_sales)


# Calculate week-on-week growth percentage

growth_percentage = ((week2_total - week1_total) / week1_total) * 100


# Display results

print("WEEKLY SALES PERFORMANCE")
print("-" * 35)

print(f"Week 1 Sales: KES {week1_total:,.2f}")
print(f"Week 2 Sales: KES {week2_total:,.2f}")
print(f"Week-on-Week Growth: {growth_percentage:.2f}%")
```

## Task 3: Daily Sales Classification

In this task, I create a function called classify_day() to classify each day's sales as Strong, Normal or Weak based on the overall average daily sales. I use a dictionary to count the number of days in each category.

```python
# Function to classify daily sales

def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal or Weak.
    """
    if amount > average * 1.20:
        return "Strong"

    elif amount < average * 0.80:
        return "Weak"

    else:
        return "Normal"


# Create a dictionary to count the categories

performance_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}


# Classify each day and count the results

daily_classifications = []

for amount in daily_sales:

    category = classify_day(amount, average_sales)

    daily_classifications.append(category)

    performance_counts[category] += 1


# Display the results

print("DAILY PERFORMANCE CLASSIFICATION")
print("-" * 40)

for category, count in performance_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday Versus Weekend Performance

In this task, I separate weekday sales from weekend sales. I calculate the average sales for each group and determine the percentage difference to identify which period performs better.

```python
# Create lists for weekday and weekend sales

weekday_sales = []

weekend_sales = []


# Separate sales according to the day

for i in range(len(daily_sales)):

    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:

        weekday_sales.append(daily_sales[i])

    else:

        weekend_sales.append(daily_sales[i])


# Calculate average sales

weekday_average = sum(weekday_sales) / len(weekday_sales)

weekend_average = sum(weekend_sales) / len(weekend_sales)


# Calculate percentage difference

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100


# Display results

print("WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 40)

print(f"Average Weekday Sales: KES {weekday_average:,.2f}")

print(f"Average Weekend Sales: KES {weekend_average:,.2f}")

print(f"Weekend Sales Difference: {percentage_difference:.2f}%")
```

## Task 5: Stock Reordering and Gross Margins

In this task, I create a reusable function that identifies products whose current stock quantity is at or below their reorder level. I also calculate the gross margin percentage for every product using its selling price and cost price.

```python
# Function to identify products requiring restocking

def get_reorder_products(stock_data):
    """
    Return a list of products whose quantity is at or below
    their specified reorder level.
    """
    reorder_products = []

    for product, details in stock_data.items():

        if details["qty"] <= details["reorder"]:

            reorder_products.append(product)

    return reorder_products


# Identify products requiring restocking

products_to_reorder = get_reorder_products(stock)


# Calculate gross margins

gross_margins = {}

for product, details in stock.items():

    cost = details["cost"]

    price = details["price"]

    margin = ((price - cost) / price) * 100

    gross_margins[product] = margin


# Display stock analysis

print("STOCK AND GROSS MARGIN ANALYSIS")
print("-" * 55)

print(f"{'Product':<15}{'Stock (kg)':<12}{'Reorder (kg)':<15}{'Margin':>10}")

print("-" * 55)

for product, details in stock.items():

    print(
        f"{product:<15}"
        f"{details['qty']:<12}"
        f"{details['reorder']:<15}"
        f"{gross_margins[product]:>9.2f}%"
    )


print("\nPRODUCTS REQUIRING RESTOCKING")
print("-" * 35)

for product in products_to_reorder:

    print(product)
```

## Task 6: Stall Performance Report

In this task, I consolidate the findings from the previous tasks into a formatted business performance report. I use Python f-strings to display financial figures with KES currency formatting and thousands separators.

```python
# Print the final business performance report

print("=" * 60)
print("             STALL PERFORMANCE REPORT")
print("             GIKOMBA MARKET")
print("=" * 60)

print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 60)

print(f"Total Sales (14 Days):       KES {total_sales:,.2f}")
print(f"Average Daily Sales:         KES {average_sales:,.2f}")
print(f"Highest Daily Sales:         KES {highest_sales:,.2f}")
print(f"Lowest Daily Sales:          KES {lowest_sales:,.2f}")

print(f"Best Sales Day:              {best_day}, Week {best_week}")
print(f"Worst Sales Day:             {worst_day}, Week {worst_week}")


print("\n2. WEEKLY SALES PERFORMANCE")
print("-" * 60)

print(f"Week 1 Sales:                KES {week1_total:,.2f}")
print(f"Week 2 Sales:                KES {week2_total:,.2f}")
print(f"Week-on-Week Growth:         {growth_percentage:.2f}%")


print("\n3. DAILY PERFORMANCE CLASSIFICATION")
print("-" * 60)

for category, count in performance_counts.items():

    print(f"{category:<15}: {count} days")


print("\n4. WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 60)

print(f"Average Weekday Sales:       KES {weekday_average:,.2f}")
print(f"Average Weekend Sales:       KES {weekend_average:,.2f}")
print(f"Weekend Advantage:           {percentage_difference:.2f}%")


print("\n5. STOCK REORDER REQUIREMENTS")
print("-" * 60)

for product in products_to_reorder:

    print(
        f"{product}: {stock[product]['qty']} kg "
        f"(Reorder level: {stock[product]['reorder']} kg)"
    )


print("\n6. PRODUCT GROSS MARGINS")
print("-" * 60)

for product, margin in gross_margins.items():

    print(f"{product:<15}: {margin:.2f}%")


print("\n" + "=" * 60)
print("             END OF PERFORMANCE REPORT")
print("=" * 60)
```
